```python
from pymongo import MongoClient
from pymongo.errors import PyMongoError
import urllib.parse

mongo_config = {
    "UseSRV": False,   # False => mongodb:// | True => mongodb+srv://
    "Host": "localhost",
    "Port": 27017,
    "DBName": "edu_chatbot",
    "Username": "",
    "Password": "",
    "AuthSource": "",
    "TLS": False,
    "TLSCAFile": ""
}

# --- Build MongoDB URI from config ---
scheme = "mongodb+srv" if mongo_config["UseSRV"] else "mongodb"

user = mongo_config.get("Username") or ""
pwd = mongo_config.get("Password") or ""

auth_part = ""
if user:
    auth_part = f"{urllib.parse.quote_plus(user)}:{urllib.parse.quote_plus(pwd)}@"

host = mongo_config["Host"]
port = mongo_config["Port"]

# For SRV: no port in URI
host_part = host if mongo_config["UseSRV"] else f"{host}:{port}"

dbname = mongo_config["DBName"]

params = []

# authSource (optional)
auth_source = mongo_config.get("AuthSource")
if auth_source:
    params.append(f"authSource={urllib.parse.quote_plus(auth_source)}")

# TLS
if mongo_config.get("TLS"):
    params.append("tls=true")
    ca_file = mongo_config.get("TLSCAFile")
    if ca_file:
        params.append(f"tlsCAFile={urllib.parse.quote_plus(ca_file)}")

uri = f"{scheme}://{auth_part}{host_part}/{dbname}"
if params:
    uri += "?" + "&".join(params)

# --- Test connection ---
try:
    client = MongoClient(uri, serverSelectionTimeoutMS=5000)
    client.admin.command("ping")  # forces real connection + auth check
    print("✅ MongoDB CONNECTED")
except PyMongoError as e:
    print("❌ MongoDB NOT CONNECTED")
    print("Error:", e)
finally:
    try:
        client.close()
    except:
        pass
```

```python
mongo_documentdb_config = {
    # URI Scheme
    # mongodb:// (Amazon DocumentDB does NOT support mongodb+srv://)
    # Equivalent: URI Scheme
    # 🔴 Mandatory
    "UseSRV": False,   # Always False for Amazon DocumentDB

    # Amazon DocumentDB cluster endpoint
    # Equivalent: Host
    # 🔴 Mandatory
    "Host": "docdb-cluster.cluster-xxxx.us-east-1.docdb.amazonaws.com",

    # MongoDB port number
    # Equivalent: Port
    # 🔴 Mandatory (DocumentDB uses fixed port 27017)
    "Port": 27017,

    # Database the application will connect to
    # Equivalent: Database Name
    # 🔴 Mandatory
    "DBName": "chatbot_db",

    # Amazon DocumentDB username
    # Equivalent: Username
    # 🔴 Mandatory
    "Username": "docdb_user",

    # Amazon DocumentDB user password
    # Equivalent: Password
    # 🔴 Mandatory
    "Password": "********",

    # Authentication database
    # Equivalent: Auth Source
    # 🟢 Optional (default = "admin")
    "AuthSource": "admin",

    # Authentication mechanism
    # Equivalent: Auth Mechanism
    # 🟢 Optional (SCRAM-SHA-256 supported)
    "AuthMechanism": "SCRAM-SHA-256",

    # Enable encrypted connection
    # Equivalent: TLS / SSL
    # 🔴 Mandatory (DocumentDB requires TLS)
    "TLS": True,

    # Path to Certificate Authority file
    # Equivalent: CA Certificate Path
    # 🔴 Mandatory (AWS provides a custom CA bundle)
    "TLSCAFile": "/etc/ssl/certs/rds-combined-ca-bundle.pem",

    # Retryable writes
    # Equivalent: Retry Writes
    # 🟢 Optional (often set to false for safety)
    "RetryWrites": False
}
```
